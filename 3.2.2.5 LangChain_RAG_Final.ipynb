# Load the RAG data

Our dataset consists of 10 documents, each containing cleaned text from a Wikipedia article on a different country. It also includes a (.csv) file containing question and answer pairs based only on these documents.

This dataset is a processed subset of a larger one that you can find linked below:


*Smith, N. A., Heilman, M., & Hwa, R. (2008, September) [[link]](https://https://www.kaggle.com/datasets/rtatman/questionanswer-dataset/data)*

Using the the code below, you can download the dataset from the url and unzip it. The list `articles_filenames` contrains the path for each (.txt) article file and `qa_eval_filename` contains the path to the (.csv) Q&A file.

```python
# Download, and extract the database
import io       # For creating an in-memory binary stream
import zipfile  # For handling zip files
import requests # For loading the file from the web

# The database is a .zip file containing .txt articles
url = "https://drive.google.com/uc?export=view&id=1Icqw0yYzmsT0CyrcTJ3UNMY9icnOQcNI"

try:
    db_zip = requests.get(url, timeout=60)
    db_zip.raise_for_status() # fail fast on bad status

    filenames = list()

    with zipfile.ZipFile(io.BytesIO(db_zip.content)) as zf:
        filenames = zf.namelist()
        zf.extractall() # extract all files in the local directory

    print(f"Downloaded {len(filenames)} files.")

except requests.exceptions.RequestException as e:
    print(f"Error downloading the file: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

# Separate eval data and articles
articles_filenames = filenames[-10:] # Feel free to limit to fewer files if it takes too long
qa_eval_filename =  filenames[0]
```

`text_splitter` is a `RecursiveCharacterTextSplitter`. We usually set `chunk_overlap` to be between 10-20% of `chink_size`.

This new version of `extract_sentences` uses langchain's TextLoader, to load each file in a `Document` and set its `source` and `subject` metadata.

Finally, `text_splitter.split_documents()` split the document into chunks recursively.

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,  # chunk size (characters)
    chunk_overlap=100,  # chunk overlap (characters)
    add_start_index=True,  # track index in original document
)

def extract_sentences(filename):
    loader = TextLoader(filename, encoding='utf-8')
    doc = loader.load()
    doc[0].metadata["source"] = filename
    doc[0].metadata["subject"] = doc[0].page_content.split('\n', 1)[0]
    print(f"Split {filename} into chunks.")

    return text_splitter.split_documents(doc)
```

# Models and Vector Store Initialization

With Ollama, we used to pass the model (downloaded via the `pull` command) name using methods like `ollama.embed()` and `ollama.generate()`. If we ever want to use a different model outside of Ollama (e.g. OpenAI models via API), we would need to rewrite everything to work with the new model's API.

One of LangChain's core components is Models. Models are object classes that allow us to encapsulate our model, so that the pipeline operates the same way regardless of where the model's from. This way we can easily switch between different models by only changing the initialization.

On top of this, LangChain offers a variety of integrations from different model providers such as Ollama, OpenAI, HuggingFace and more. You can find all supported providers [here](https://docs.langchain.com/oss/python/integrations/providers/all_providers).

For Ollama, we have `ChatOllama()` and `OllamaEmbeddings()` which we can import form `langchain_ollama`.

```python
# Set up the Ollama models
from langchain_ollama import ChatOllama

# Uncomment and use if you prefer the Gemma setup
# llm_name = "gemma3:1b"
# embed_name = "embeddinggemma:300m"

# Make sure the names here match the models you've pulled and want to use
llm_name = "deepseek-r1:8b"
embed_name = "bge-large:335m"

# We can still use Gemma for eval to keep the results consistent
# And also have a faster evaluation
# You can change this to deepseek-r1 or any llm of your choosing
eval_name = "gemma3:1b"
# eval_name = "deepseek-r1:8b"

llm_model = ChatOllama(model=llm_name,
                 temprature=0)
eval_model = ChatOllama(model=eval_name,
                 temprature=0)

# Set up the Ollama embedding model
from langchain_ollama import OllamaEmbeddings

embedding_model = OllamaEmbeddings(model=embed_name)
```

Next, let's define the vector store. This is another core component of LangChain. For now we use the `InMemoryVectorStore` from `langchain_core.vectorstores`. The vector store object initializer, expects an embedding model as input. This model is used to vectorize any chunk that is added to the vector store.

As with the model class, we can easily change our vector store by using a different initializer. We'll explore vectors further in the next video.

```python
# Set up vector storing in memory
from langchain_core.vectorstores import InMemoryVectorStore

vector_store = InMemoryVectorStore(embedding_model)
```

# Populate the Vector Store

Now, we chunk the data and add it to `vector_store`. The vector store automatically embeds the chunks using the embedding model.

## LangChain `Document`

An important object in LangChain is a `Document`. Previously, we use this only for the source documents in the dataset.
However, in LangChain, `Document` is an abstraction that represents a unit of text and its associated metadata. Often, this refers to a chunk of text.

A `Document` has 3 attributes:


*   `page_content: str`
*   `metadata: dict`
*   `id (optional): str`

We'll use `metadata` to track the information we previously stored in `Chunk`.



```python
from importlib.metadata import metadata

all_docs = list()
for filename in articles_filenames:
    all_docs.extend(extract_sentences(filename))

_ = vector_store.add_documents(all_docs)
print(f"Added {len(all_docs)} documents to the vector store.")
```

# Create LangChain `Retriever`

In this version, you can choose from 3 different setups:


1.   Retriever -> Reranker
2.   Retriever -> Rephraser
3.   Retriever -> Rephraser -> Reranker

*Not that the setups are in increasing order of computational cost.*

***Make sure to only run the cell corresponding to your chosen setup. You can comment out the rest.***

## 1. Retriever -> Reranker

```python
!pip install -qU  flashrank

from flashrank import Ranker, RerankRequest # required for FlashrankRerank
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_community.document_compressors import FlashrankRerank

ranker = Ranker(model_name="ms-marco-MiniLM-L-12-v2")

# reranker
reranker = FlashrankRerank(
    client=ranker,
    top_n=5 # Adjust as needed
)

# base retriever from vector store
base_retriever = vector_store.as_retriever(search_type="similarity",
                                      search_kwargs={"k": 20}) # Adjust as needed

# Combine base + reranker
retriever = ContextualCompressionRetriever(
    base_compressor=reranker,
    base_retriever=base_retriever
)
```

## 2. Retriever -> Rephraser

```python
from langchain_classic.retrievers import RePhraseQueryRetriever

# Optional logging
# import logging
# logging.basicConfig()
# logging.getLogger("langchain_classic.retrievers").setLevel(logging.INFO)

# base retriever from vector store
base_retriever = vector_store.as_retriever(search_type="similarity",
                                      search_kwargs={"k": 5}) # Adjust as needed

# Combine base + rephraser
retriever = RePhraseQueryRetriever.from_llm(
    retriever=base_retriever, llm=llm_model
)
```

## 3. Retriever -> Rephraser -> Reranker

```python
!pip install -qU  flashrank

from flashrank import Ranker, RerankRequest # required for FlashrankRerank
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_community.document_compressors import FlashrankRerank

from langchain_classic.retrievers import RePhraseQueryRetriever

# Optional logging
# import logging
# logging.basicConfig()
# logging.getLogger("langchain_classic.retrievers").setLevel(logging.INFO)

ranker = Ranker(model_name="ms-marco-MiniLM-L-12-v2")

# reranker
reranker = FlashrankRerank(
    client=ranker,
    top_n=5 # Adjust as needed
)

# base retriever from vector store
base_retriever_1 = vector_store.as_retriever(search_type="similarity",
                                      search_kwargs={"k": 20}) # Adjust as needed

# Combine base + rephraser
base_retriever_2 = RePhraseQueryRetriever.from_llm(
    retriever=base_retriever, llm=llm_model # You could use a smaller/faster LLM for the RePhraser
)

# Combine (base + rephraser) + reranker
retriever = ContextualCompressionRetriever(
    base_compressor=reranker,
    base_retriever=base_retriever_2
)
```

# Define Prompt Templates

```python
from langchain_core.prompts import ChatPromptTemplate

SYSTEM_PROMPT = ("You are a helpful question-answering chatbot. \n"
                  "Use only the following pieces of context to answer the question. \n"
                  "If you don't know the answer, just say that you don't know and don't make up any new information.")

USER_TEMPLATE = ("Here is the context: \n{context_string} \n"
                    "Question: {question}\n\n")
                    # "Answer:")

PROMPT_TEMPLATE = ChatPromptTemplate([
    ("system", SYSTEM_PROMPT),
    ("human", USER_TEMPLATE)
])
```

# Retrieve and Generate

```python
def answer_query(input_query, verbose=False):
    # Retrieve the top 5 relevant chunks
    retrieved_docs = retriever.invoke(input_query)
    # Join context chunks in one string
    context_string = '\n'.join([doc.page_content for doc in retrieved_docs])

    # Fill in the prompt templates with the question and context
    chat_prompt = PROMPT_TEMPLATE.format_messages(question=input_query,
                                         context_string=context_string)

    # Generate response
    response = llm_model.invoke(chat_prompt)

    if verbose:
        print("Chatbot response:")
        print(response.content)
        print("------------------------------------------")
        print(f'Retrieved context:')
        for n, doc in enumerate(retrieved_docs):
            content = doc.page_content.replace('\n', ' ')
            print(f"* {n+1} \t Subject: {doc.metadata['subject']} \n{content}\n")
    else:
        return response.content, retrieved_docs
```

## Correctness Evaluation Using LLM-as-Judge

```python
from typing_extensions import Annotated, TypedDict

# Grade output schema
class CorrectnessGrade(TypedDict):
    # It's important to order explaination first, followed by score
    # Since it forces the model to think its final grade through
    explanation: Annotated[str, "Explain the reasoning for your answered score."]
    correctness: Annotated[int, "Grade tha answers correctness on a 0-5 scale."]

CORRECTNESS_PROMPT = """You are a careful evaluator. You will be given a QUESTION,
the GROUND TRUTH (correct) ANSWER, and a STUDENT'S ANSWER. You will grade the
student's CORRECTNESS based on the following criteria:

(1) Grade the student's answer based ONLY on its factual accuracy relative to
the ground truth answer.
(2) It is OK if the student's answer is less concise, as long as it is semantically correct.
(2) The student's answer should not contain any contradicting statements.
(4) Even if the student's answer is partially similar to the ground truth,
it still must fully address the question asked.

Correctness rubric:
- 5: (Exactly Correct): Fully aligns with Ground Truth; no material error/omission.
- 4: (Very Good): Correct overall; tiny omission or benign imprecision that doesn’t change meaning.
- 3: (Fair/Partially Correct): Core idea present but with notable omission or one substantive error.
- 2: (Limited): Some correct fragments, but overall meaning materially off.
- 1: (Poor): Barely overlaps; main claim incorrect.
- 0: (Incorrect/No Answer): Completely wrong, or irrelevant.

Instead of declaring the grade outright, first, explain your reasoning for the
grade, step-by-step, and then decide the grade based on this reasoning."""


def evaluate_correctness(llm, question, ground_truth, student_answer):
    grader_llm = llm.with_structured_output(CorrectnessGrade,
                                            method="json_schema",
                                            strict=True)

    TESTCASE_TEMPLATE = ("QUESTION: \n{question} \n\n"
                        "-------------------------------------\n"
                        "GROUND TRUTH: \n{ground_truth}\n\n"
                        "-------------------------------------\n"
                        "STUDENT'S ANSWER: \n{student_answer}")

    PROMPT = ChatPromptTemplate.from_messages([
        ("system", CORRECTNESS_PROMPT),
        ("human", TESTCASE_TEMPLATE)
    ]).format_messages(question=question,
                      ground_truth=ground_truth,
                      student_answer=student_answer)

    response = grader_llm.invoke(PROMPT)
    return response
```

## Evaluate Correctness Using Q&A Dataset

```python
from tqdm import tqdm

RAG_eval_results = {"correctness": 0.0}

nc, nr = 0, 0
for i in tqdm(range(len(df))):
    row = df.iloc[i]

    question = row['Question']
    ground_truth = row['Answer']
    student_answer, retrieved_context = answer_query(question, verbose=False)

    try:
        # You could use a smaller/faster LLM as judge for the function call
        correctness = evaluate_correctness(eval_model, question, ground_truth, student_answer)["correctness"]
        if correctness in [0,1,2,3,4,5]:
            nc += 1
            RAG_eval_results["correctness"] += correctness
    except Exception as e:
        print(f"Error evaluating correctness for question {i+1}: {e}")

RAG_eval_results["correctness"] /= nc
```

## RAGAS Evaluation

Metrics we'll evaluate with RAGAS:

*   LLMContextRecall()
*   AnswerRelevancy()
*   Faithfulness()
*   FactualCorrectness() : Calculates an F1 score

See for more on RAGAS metrics: https://docs.ragas.io/en/stable/concepts/metrics/available_metrics/

```python
!pip install ragas
```

```python
from ragas import evaluate
from datasets import Dataset
from ragas.run_config import RunConfig
from ragas.metrics import LLMContextRecall, AnswerRelevancy, Faithfulness, FactualCorrectness


# Prepare the data for RAGAS evaluation
# We need to map our DataFrame columns to the expected RAGAS format
# 'question', 'ground_truth', 'answer'
# 'answer' will be the output of our RAG pipeline
ragas_data = []
for i in tqdm(range(len(df))):
    row = df.iloc[i]
    question = row['Question']
    ground_truth = row['Answer']
    # Get the RAG answer for the question and the retrieved documents
    rag_answer, retrieved_docs = answer_query(question)
    ragas_data.append({
        'question': question,
        'ground_truth': ground_truth,
        'answer': rag_answer,
        'retrieved_contexts': [doc.page_content for doc in retrieved_docs]
    })

# Convert the list of dictionaries to a Hugging Face Dataset
ragas_dataset = Dataset.from_list(ragas_data)

# Evaluate the RAG pipeline
# You can choose which metrics to use. Here are some common ones:
# faithfulness, answer_relevancy, context_precision, context_recall, etc.
# For a full list, see: https://docs.ragas.io/en/latest/concepts/metrics/index.html
result = evaluate(
    ragas_dataset,
    metrics=[
        LLMContextRecall(),
        AnswerRelevancy(),
        Faithfulness(),
        FactualCorrectness()
        ],
    # You can specify the LLM to use for evaluation if needed,
    llm=eval_model, # You could use a smaller/faster LLM as judge
    embeddings=embedding_model,
    run_config=RunConfig(timeout=500, max_wait=180, max_workers=64) # Optional; Adjusts run parameters
)

# Convert the result to a Pandas DataFrame for better readability
results_df = result.to_pandas()
```
